## Paso 1 — Nuestros "documentos"

En NotebookLM cargas PDFs. Acá simulamos con strings.

```python
documentos = [
    """3. Usuario / cliente objetivo
Familias, parejas, grupos de amigos y companeros de trabajo que valoran el encuentro en torno a la comida (almuerzos, cenas, la "once") pero sufren de la distraccion constante de las notificaciones, y que buscan una herramienta cotidiana y fluida para reactivar dinamicas de conversacion profunda y distension en el mundo fisico.""",

    """Cantidad de registros por ocupación:
occupation
Employed         207
Student          107
Self-employed     45
Unemployed        27
Retired           14
Name: count, dtype: int64

Bienestar mental promedio por ocupación:
occupation
Student          15.539252
Employed         20.188406
Self-employed    20.477778
Retired          31.578571
Unemployed       34.274074
Name: mental_wellness_index_0_100, dtype: float64""",



]
print(f"{len(documentos)} documentos cargados.")
```

## Paso 2 — Convertir documentos a embeddings

```python
from sentence_transformers import SentenceTransformer
import numpy as np

modelo = SentenceTransformer("sentence-transformers/all-MiniLM-L6-v2")
embeddings = modelo.encode(documentos)
print(f"Shape: {embeddings.shape}")
print("Cada documento es ahora un vector de 384 dimensiones.")
```

## Paso 3 — Función de búsqueda semántica

Le damos una pregunta, ella busca el documento más parecido.

```python
def buscar(pregunta, top_k=1):
    p_emb = modelo.encode([pregunta])
    sims = np.dot(embeddings, p_emb.T).flatten()
    sims = sims / (np.linalg.norm(embeddings, axis=1) * np.linalg.norm(p_emb))
    top_idx = np.argsort(sims)[::-1][:top_k]
    return [(documentos[i], float(sims[i])) for i in top_idx]

preguntas = [
    "cuál es el cliente objetivo?",
    "hay algo sobre el bienestar mental?",
    "cuántos estudiantes hay?",
]

for q in preguntas:
    print(f"P: {q}")
    for doc, sim in buscar(q, top_k=1):
        print(f"  Similitud: {sim:.2f}")
        print(f"  Documento: {doc[:150]}...")
    print()
```
